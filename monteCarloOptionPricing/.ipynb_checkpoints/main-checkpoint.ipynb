In finance, an option is a type of financial derivative which gives the owner the right (but not the obligation) to purchase (calls) or to sell (puts) a certain stock at a set strike price from being issued until expiration. 

The financial flexibility options provide come at a premium. Working out the correct price of an option is necessary for fair value and to manage risk. In standard pricing theory, the price of an option should reflect the discounted expected payoff under a risk-neutral measure. However, markets are often complex systems to model numerically. Black-Scholes, the standard model for pricing European options, relies on assumptions specific to Europian markets. Exotic options such as Asian options require a different approach. 

The Monte Carlo simulation provides an intuitive solution. From the starting price, in each iteration the price changes randomly under Geometric Brownian Motion (GBM). This stochastic process is parameterised to model the volatility and starting conditions of the stock, with restrictions ensuring the stock price stays positive with compounding returns. Over many interations, an average return at the expiry date will be accurately estimated, due to the Law of Large Numbers. 

This notebook compares the similarity between option prices derived through Monte Carlo and Black-Scholes. It then extends Monte Carlo to Asian options, which cannot be modelled by the Black-Scholes, demonstrating the wider breadth of Monte Carlo option pricing over Black-Scholes. Finally, antithetic-variates are analysed as a tool to reduce variance and hence reduce the high computational cost of an effective Monte Carlo simulation.

```python
#importing dependencies
import yfinance as yf
import pandas as pd
import numpy as np
from numpy import random
from scipy.stats import norm
import matplotlib.pyplot as plt
import importlib
import models as mod
importlib.reload(mod)
```

```python
price_data = yf.download(
    'SHEL.L', 
    start='2010-01-01', 
    end='2025-01-01', 
)
```

```python
#set seed for reproducibility
rng = np.random.default_rng(10)

# Define the option
trading_days = 252
S_0 = price_data["Close"].iloc[-1].to_numpy()[0] # Current stock price (taken as latest adjusted close price)
K = S_0                            # Strike price 
T = 0.5                            # Time to expiry (years)
d_t = T / trading_days             # Time step (years)
r = 0.0393                         # Risk-free interest rate

# historical drift
log_returns = np.log(
    price_data["Close"] / price_data["Close"].shift(1)
)
log_returns = log_returns.dropna()
sigma = np.sqrt(trading_days) * log_returns.std(ddof=1).to_numpy()  # Volatility (standard deviation of logged stock returns)
```

```python
BS_optionPrice = mod.Black_Scholes(S_0, r, sigma, T, K, option_type = "Put")
```

```python
simulations = 1000
euro_put_est = mod.visualise_GBM(simulations, S_0, r, sigma, T, K, d_t, market_type="European", option_type = "Put")
```

```python
asian_call_est = mod.visualise_GBM(simulations, S_0, r, sigma, T, K, d_t, market_type="Asian", option_type = "CALL")
```

Monte Carlo is flexible, but computationally expensive. Hence, I used antithetic-variates to reduce variance.

```python
antithetic_euro_call_est = mod.visualise_GBM(simulations, S_0, r, sigma, T, K, d_t, market_type="european", option_type = "call", antithetic_var=True)
BS_antithetic_call_price = mod.Black_Scholes(S_0, r, sigma, T, K, option_type="call")
```
